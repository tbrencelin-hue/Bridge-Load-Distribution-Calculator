```python
# ============================================================
# BRIDGE LOAD DISTRIBUTION CALCULATOR
# B.Tech Mathematics Mini Project
# Google Colab + Gradio
# ============================================================

!pip -q install gradio


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import gradio as gr


# ============================================================
# 1. MATHEMATICAL ENGINE
# ============================================================

def solve_bridge(
    xB,
    xC,
    p1_pos,
    p1_load,
    p2_pos,
    p2_load,
    p3_pos,
    p3_load
):

    # --------------------------------------------------------
    # Basic input validation
    # --------------------------------------------------------

    if xB <= 0:
        raise ValueError("Support B position must be greater than 0.")

    if xC <= xB:
        raise ValueError(
            "Support C must be located to the right of Support B."
        )

    # --------------------------------------------------------
    # Load positions and loads
    # --------------------------------------------------------

    positions = np.array([
        p1_pos,
        p2_pos,
        p3_pos
    ], dtype=float)

    loads = np.array([
        p1_load,
        p2_load,
        p3_load
    ], dtype=float)

    # --------------------------------------------------------
    # MATRIX A
    #
    # Unknowns:
    #
    # X = [RA, RB, RC]^T
    #
    # Equation 1:
    # RA + RB + RC = total load
    #
    # Equation 2:
    # xB*RB + xC*RC = total moment
    #
    # Equation 3:
    # RC = RA
    #
    # Therefore:
    # -RA + RC = 0
    # --------------------------------------------------------

    A = np.array([
        [1, 1, 1],
        [0, xB, xC],
        [-1, 0, 1]
    ], dtype=float)

    # --------------------------------------------------------
    # MATRIX B
    # --------------------------------------------------------

    total_load = np.sum(loads)

    total_moment = np.sum(
        loads * positions
    )

    B = np.array([
        total_load,
        total_moment,
        0
    ], dtype=float)

    # --------------------------------------------------------
    # CHECK DETERMINANT
    # --------------------------------------------------------

    determinant = np.linalg.det(A)

    if abs(determinant) < 1e-10:
        raise ValueError(
            f"Matrix A is singular.\n"
            f"det(A) = {determinant}\n\n"
            "Check the support positions."
        )

    # --------------------------------------------------------
    # MATRIX INVERSE
    # --------------------------------------------------------

    A_inverse = np.linalg.inv(A)

    # --------------------------------------------------------
    # SOLUTION
    #
    # X = A^-1 B
    # --------------------------------------------------------

    X = A_inverse @ B

    # --------------------------------------------------------
    # VERIFY AX = B
    # --------------------------------------------------------

    calculated_B = A @ X

    error = np.max(
        np.abs(calculated_B - B)
    )

    return A, B, A_inverse, X, error


# ============================================================
# 2. BRIDGE DIAGRAM
# ============================================================

def bridge_diagram(
    xB,
    xC,
    positions,
    loads,
    truck_position
):

    fig, ax = plt.subplots(
        figsize=(11, 4)
    )

    # Bridge
    ax.plot(
        [0, xC],
        [0, 0],
        linewidth=6
    )

    # Supports
    ax.scatter(
        [0, xB, xC],
        [0, 0, 0],
        marker="^",
        s=180
    )

    # Support labels
    ax.text(
        0,
        -0.35,
        "A",
        ha="center",
        fontsize=13,
        fontweight="bold"
    )

    ax.text(
        xB,
        -0.35,
        "B",
        ha="center",
        fontsize=13,
        fontweight="bold"
    )

    ax.text(
        xC,
        -0.35,
        "C",
        ha="center",
        fontsize=13,
        fontweight="bold"
    )

    # Loads
    for i, (pos, load) in enumerate(
        zip(positions, loads)
    ):

        ax.arrow(
            pos,
            2.2,
            0,
            -1.7,
            head_width=0.2,
            head_length=0.2,
            length_includes_head=True
        )

        ax.text(
            pos,
            2.35,
            f"P{i+1} = {load:.0f} kN",
            ha="center"
        )

    # Truck
    ax.text(
        truck_position,
        0.65,
        "🚚",
        fontsize=24,
        ha="center"
    )

    ax.set_title(
        "Bridge Load Distribution",
        fontsize=15,
        fontweight="bold"
    )

    ax.set_xlabel(
        "Distance along bridge (m)"
    )

    ax.set_xlim(
        -1,
        xC + 1
    )

    ax.set_ylim(
        -0.7,
        3
    )

    ax.grid(
        alpha=0.25
    )

    return fig


# ============================================================
# 3. REACTION COMPARISON
# ============================================================

def comparison_chart(
    before,
    after
):

    supports = [
        "A",
        "B",
        "C"
    ]

    x = np.arange(3)

    width = 0.35

    fig, ax = plt.subplots(
        figsize=(9, 5)
    )

    ax.bar(
        x - width/2,
        before,
        width,
        label="Before"
    )

    ax.bar(
        x + width/2,
        after,
        width,
        label="After"
    )

    ax.set_xticks(x)

    ax.set_xticklabels(
        supports
    )

    ax.set_ylabel(
        "Reaction Force (kN)"
    )

    ax.set_xlabel(
        "Bridge Support"
    )

    ax.set_title(
        "Support Reaction Comparison"
    )

    ax.legend()

    ax.grid(
        axis="y",
        alpha=0.3
    )

    return fig


# ============================================================
# 4. MAIN CALCULATION
# ============================================================

def calculate(
    xB,
    xC,
    p1_pos,
    p1_load,
    p2_pos,
    p2_load,
    p3_pos,
    p3_load,
    truck_position
):

    try:

        # ----------------------------------------------------
        # ORIGINAL SOLUTION
        # ----------------------------------------------------

        A, B, A_inv, before, error = solve_bridge(
            xB,
            xC,
            p1_pos,
            p1_load,
            p2_pos,
            p2_load,
            p3_pos,
            p3_load
        )

        # ----------------------------------------------------
        # MOVE P2
        # ----------------------------------------------------

        A2, B2, A_inv2, after, error2 = solve_bridge(
            xB,
            xC,
            p1_pos,
            p1_load,
            truck_position,
            p2_load,
            p3_pos,
            p3_load
        )

        # ----------------------------------------------------
        # MATRIX TEXT
        # ----------------------------------------------------

        A_text = np.array2string(
            A,
            precision=3,
            suppress_small=True
        )

        B_text = np.array2string(
            B,
            precision=3,
            suppress_small=True
        )

        inverse_text = np.array2string(
            A_inv,
            precision=4,
            suppress_small=True
        )

        # ----------------------------------------------------
        # RESULTS TABLE
        # ----------------------------------------------------

        result_table = pd.DataFrame({

            "Support": [
                "A",
                "B",
                "C"
            ],

            "Before (kN)": np.round(
                before,
                3
            ),

            "After (kN)": np.round(
                after,
                3
            ),

            "Change (kN)": np.round(
                after - before,
                3
            )
        })

        # ----------------------------------------------------
        # PLOTS
        # ----------------------------------------------------

        positions = np.array([
            p1_pos,
            p2_pos,
            p3_pos
        ])

        loads = np.array([
            p1_load,
            p2_load,
            p3_load
        ])

        bridge_fig = bridge_diagram(
            xB,
            xC,
            positions,
            loads,
            truck_position
        )

        comparison_fig = comparison_chart(
            before,
            after
        )

        # ----------------------------------------------------
        # SUMMARY
        # ----------------------------------------------------

        summary = f"""
# ✅ Calculation Successful

## Mathematical Model

**AX = B**

### Determinant

**det(A) = {np.linalg.det(A):.4f}**

Since det(A) ≠ 0, matrix A is invertible.

Therefore, the system has a **unique solution**.

---

## Before Truck Movement

**Rₐ = {before[0]:.3f} kN**

**Rᵦ = {before[1]:.3f} kN**

**R𝒸 = {before[2]:.3f} kN**

---

## After Truck Movement

Truck moved to:

**{truck_position:.2f} m**

**Rₐ = {after[0]:.3f} kN**

**Rᵦ = {after[1]:.3f} kN**

**R𝒸 = {after[2]:.3f} kN**

---

## Verification

Maximum error in AX = B:

**{error:.10e}**

If this value is approximately zero, the solution is verified.
"""

        return (
            summary,
            A_text,
            B_text,
            inverse_text,
            result_table,
            bridge_fig,
            comparison_fig
        )

    except Exception as e:

        return (
            f"""
# ❌ Input Error

**{str(e)}**

Please check the bridge support positions and load values.
""",
            "",
            "",
            "",
            pd.DataFrame(),
            None,
            None
        )


# ============================================================
# 5. GRADIO FRONTEND
# ============================================================

with gr.Blocks(
    title="Bridge Load Distribution Calculator"
) as demo:

    gr.Markdown(
        """
# 🏗️ Bridge Load Distribution Calculator

### B.Tech Mathematics Mini Project — A2

**AX = B | Matrix Inverse | Unique Solution**

Calculate the reaction forces at three bridge supports
and observe how the reactions change when a truck moves.
"""
    )

    # --------------------------------------------------------
    # INPUTS
    # --------------------------------------------------------

    gr.Markdown(
        "## 🌉 Bridge Configuration"
    )

    with gr.Row():

        xB = gr.Number(
            value=5,
            label="Support B Position (m)"
        )

        xC = gr.Number(
            value=10,
            label="Support C Position (m)"
        )

    gr.Markdown(
        "## 📦 Load 1"
    )

    with gr.Row():

        p1_pos = gr.Number(
            value=2,
            label="P1 Position (m)"
        )

        p1_load = gr.Number(
            value=20,
            label="P1 Load (kN)"
        )

    gr.Markdown(
        "## 📦 Load 2 / Truck"
    )

    with gr.Row():

        p2_pos = gr.Number(
            value=5,
            label="P2 Initial Position (m)"
        )

        p2_load = gr.Number(
            value=30,
            label="P2 Load (kN)"
        )

    gr.Markdown(
        "## 📦 Load 3"
    )

    with gr.Row():

        p3_pos = gr.Number(
            value=8,
            label="P3 Position (m)"
        )

        p3_load = gr.Number(
            value=25,
            label="P3 Load (kN)"
        )

    # --------------------------------------------------------
    # TRUCK SLIDER
    # --------------------------------------------------------

    gr.Markdown(
        "## 🚚 Move the Truck"
    )

    truck_position = gr.Slider(
        minimum=1,
        maximum=9,
        value=7,
        step=0.5,
        label="New P2 / Truck Position (m)"
    )

    calculate_button = gr.Button(
        "🔵 CALCULATE",
        variant="primary"
    )

    # --------------------------------------------------------
    # OUTPUT
    # --------------------------------------------------------

    gr.Markdown(
        "---"
    )

    summary = gr.Markdown()

    with gr.Row():

        A_output = gr.Textbox(
            label="Matrix A",
            lines=5
        )

        B_output = gr.Textbox(
            label="Vector B",
            lines=5
        )

    inverse_output = gr.Textbox(
        label="Matrix A⁻¹",
        lines=8
    )

    result_table = gr.Dataframe(
        label="Reaction Forces",
        interactive=False
    )

    # --------------------------------------------------------
    # CHARTS
    # --------------------------------------------------------

    gr.Markdown(
        "## 📊 Visual Results"
    )

    bridge_plot = gr.Plot(
        label="Bridge Diagram"
    )

    comparison_plot = gr.Plot(
        label="Before vs After"
    )

    # --------------------------------------------------------
    # BUTTON
    # --------------------------------------------------------

    calculate_button.click(

        fn=calculate,

        inputs=[
            xB,
            xC,

            p1_pos,
            p1_load,

            p2_pos,
            p2_load,

            p3_pos,
            p3_load,

            truck_position
        ],

        outputs=[
            summary,
            A_output,
            B_output,
            inverse_output,
            result_table,
            bridge_plot,
            comparison_plot
        ]
    )


# ============================================================
# 6. START APPLICATION
# ============================================================

demo.launch(
    share=True
)
```

```python
# ============================================================
# BRIDGE LOAD DISTRIBUTION CALCULATOR
# B.Tech Mathematics Mini Project - A2
# Google Colab Frontend
# ============================================================

!pip -q install gradio


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import gradio as gr


# ============================================================
# MATHEMATICAL ENGINE
# ============================================================

def solve_bridge(
    xA,
    xB,
    xC,
    p1_pos,
    p1_load,
    p2_pos,
    p2_load,
    p3_pos,
    p3_load
):

    # --------------------------------------------------------
    # INPUT VALIDATION
    # --------------------------------------------------------

    if xA < 0:
        raise ValueError("Support A position cannot be negative.")

    if xB <= xA:
        raise ValueError(
            "Support B must be to the right of Support A."
        )

    if xC <= xB:
        raise ValueError(
            "Support C must be to the right of Support B."
        )

    # --------------------------------------------------------
    # LOAD DATA
    # --------------------------------------------------------

    positions = np.array([
        p1_pos,
        p2_pos,
        p3_pos
    ], dtype=float)

    loads = np.array([
        p1_load,
        p2_load,
        p3_load
    ], dtype=float)

    # --------------------------------------------------------
    # MATRIX A
    #
    # Unknowns:
    #
    # X = [RA, RB, RC]^T
    #
    # Equation 1:
    # RA + RB + RC = P1 + P2 + P3
    #
    # Equation 2:
    # (xB-xA)RB + (xC-xA)RC
    # =
    # P1(xP1-xA) + P2(xP2-xA) + P3(xP3-xA)
    #
    # Equation 3:
    # RC = RA
    #
    # Therefore:
    # -RA + RC = 0
    # --------------------------------------------------------

    A = np.array([
        [1, 1, 1],
        [0, xB - xA, xC - xA],
        [-1, 0, 1]
    ], dtype=float)

    # --------------------------------------------------------
    # VECTOR B
    # --------------------------------------------------------

    total_load = np.sum(loads)

    total_moment = np.sum(
        loads * (positions - xA)
    )

    B = np.array([
        total_load,
        total_moment,
        0
    ], dtype=float)

    # --------------------------------------------------------
    # DETERMINANT
    # --------------------------------------------------------

    determinant = np.linalg.det(A)

    if abs(determinant) < 1e-10:
        raise ValueError(
            "Matrix A is singular. Please check the support positions."
        )

    # --------------------------------------------------------
    # MATRIX INVERSE
    # --------------------------------------------------------

    A_inverse = np.linalg.inv(A)

    # --------------------------------------------------------
    # SOLVE AX = B
    #
    # X = A^-1 B
    # --------------------------------------------------------

    X = A_inverse @ B

    # --------------------------------------------------------
    # VERIFY
    # --------------------------------------------------------

    calculated_B = A @ X

    error = np.max(
        np.abs(calculated_B - B)
    )

    return (
        A,
        B,
        A_inverse,
        X,
        error
    )


# ============================================================
# BRIDGE DIAGRAM
# ============================================================

def create_bridge_plot(
    xA,
    xB,
    xC,
    positions,
    loads,
    truck_position
):

    fig, ax = plt.subplots(
        figsize=(12, 4)
    )

    # Bridge
    ax.plot(
        [xA, xC],
        [0, 0],
        linewidth=6
    )

    # Supports A, B, C
    ax.scatter(
        [xA, xB, xC],
        [0, 0, 0],
        marker="^",
        s=200
    )

    # Support labels
    ax.text(
        xA,
        -0.35,
        "A",
        ha="center",
        fontsize=13,
        fontweight="bold"
    )

    ax.text(
        xB,
        -0.35,
        "B",
        ha="center",
        fontsize=13,
        fontweight="bold"
    )

    ax.text(
        xC,
        -0.35,
        "C",
        ha="center",
        fontsize=13,
        fontweight="bold"
    )

    # Loads
    for i, (pos, load) in enumerate(
        zip(positions, loads)
    ):

        ax.arrow(
            pos,
            2.2,
            0,
            -1.7,
            head_width=0.18,
            head_length=0.20,
            length_includes_head=True
        )

        ax.text(
            pos,
            2.35,
            f"P{i+1} = {load:.0f} kN",
            ha="center",
            fontsize=10
        )

    # Truck
    ax.text(
        truck_position,
        0.7,
        "🚚",
        fontsize=25,
        ha="center"
    )

    ax.set_title(
        "Bridge Load Distribution",
        fontsize=15,
        fontweight="bold"
    )

    ax.set_xlabel(
        "Distance along bridge (m)"
    )

    ax.set_xlim(
        xA - 1,
        xC + 1
    )

    ax.set_ylim(
        -0.8,
        3
    )

    ax.grid(
        alpha=0.25
    )

    return fig


# ============================================================
# REACTION COMPARISON CHART
# ============================================================

def create_comparison_chart(
    before,
    after
):

    supports = [
        "A",
        "B",
        "C"
    ]

    x = np.arange(3)

    width = 0.35

    fig, ax = plt.subplots(
        figsize=(9, 5)
    )

    ax.bar(
        x - width / 2,
        before,
        width,
        label="Before Truck Movement"
    )

    ax.bar(
        x + width / 2,
        after,
        width,
        label="After Truck Movement"
    )

    ax.set_xticks(x)

    ax.set_xticklabels(
        supports
    )

    ax.set_xlabel(
        "Support"
    )

    ax.set_ylabel(
        "Reaction Force (kN)"
    )

    ax.set_title(
        "Support Reaction Comparison",
        fontsize=14,
        fontweight="bold"
    )

    ax.legend()

    ax.grid(
        axis="y",
        alpha=0.25
    )

    return fig


# ============================================================
# TRUCK MOVEMENT GRAPH
# ============================================================

def create_movement_chart(
    xA,
    xB,
    xC,
    p1_pos,
    p1_load,
    p2_load,
    p3_pos,
    p3_load
):

    truck_positions = np.linspace(
        xA + 0.5,
        xC - 0.5,
        25
    )

    reaction_history = []

    for truck_position in truck_positions:

        _, _, _, reaction, _ = solve_bridge(

            xA,
            xB,
            xC,

            p1_pos,
            p1_load,

            truck_position,
            p2_load,

            p3_pos,
            p3_load
        )

        reaction_history.append(
            reaction
        )

    reaction_history = np.array(
        reaction_history
    )

    fig, ax = plt.subplots(
        figsize=(10, 5)
    )

    ax.plot(
        truck_positions,
        reaction_history[:, 0],
        marker="o",
        label="R_A"
    )

    ax.plot(
        truck_positions,
        reaction_history[:, 1],
        marker="s",
        label="R_B"
    )

    ax.plot(
        truck_positions,
        reaction_history[:, 2],
        marker="^",
        label="R_C"
    )

    ax.set_xlabel(
        "Truck Position (m)"
    )

    ax.set_ylabel(
        "Reaction Force (kN)"
    )

    ax.set_title(
        "Reaction Forces vs Truck Position",
        fontsize=14,
        fontweight="bold"
    )

    ax.legend()

    ax.grid(
        alpha=0.25
    )

    return fig


# ============================================================
# MAIN CALCULATION
# ============================================================

def calculate_bridge(

    xA,
    xB,
    xC,

    p1_pos,
    p1_load,

    p2_pos,
    p2_load,

    p3_pos,
    p3_load,

    truck_position
):

    try:

        # ----------------------------------------------------
        # ORIGINAL BRIDGE
        # ----------------------------------------------------

        (
            A,
            B,
            A_inverse,
            before,
            error
        ) = solve_bridge(

            xA,
            xB,
            xC,

            p1_pos,
            p1_load,

            p2_pos,
            p2_load,

            p3_pos,
            p3_load
        )

        # ----------------------------------------------------
        # AFTER TRUCK MOVEMENT
        # ----------------------------------------------------

        (
            A2,
            B2,
            A_inverse2,
            after,
            error2
        ) = solve_bridge(

            xA,
            xB,
            xC,

            p1_pos,
            p1_load,

            truck_position,
            p2_load,

            p3_pos,
            p3_load
        )

        # ----------------------------------------------------
        # MATRIX DISPLAY
        # ----------------------------------------------------

        A_text = (
            "A =\n\n"
            + np.array2string(
                A,
                precision=3,
                suppress_small=True
            )
        )

        B_text = (
            "B =\n\n"
            + np.array2string(
                B,
                precision=3,
                suppress_small=True
            )
        )

        inverse_text = (
            "A⁻¹ =\n\n"
            + np.array2string(
                A_inverse,
                precision=4,
                suppress_small=True
            )
        )

        # ----------------------------------------------------
        # RESULTS TABLE
        # ----------------------------------------------------

        results = pd.DataFrame({

            "Support": [
                "A",
                "B",
                "C"
            ],

            "Before (kN)": np.round(
                before,
                3
            ),

            "After (kN)": np.round(
                after,
                3
            ),

            "Change (kN)": np.round(
                after - before,
                3
            )
        })

        # ----------------------------------------------------
        # LOAD DATA
        # ----------------------------------------------------

        positions = np.array([
            p1_pos,
            p2_pos,
            p3_pos
        ])

        loads = np.array([
            p1_load,
            p2_load,
            p3_load
        ])

        # ----------------------------------------------------
        # CHARTS
        # ----------------------------------------------------

        bridge_fig = create_bridge_plot(

            xA,
            xB,
            xC,

            positions,
            loads,

            truck_position
        )

        comparison_fig = create_comparison_chart(
            before,
            after
        )

        movement_fig = create_movement_chart(

            xA,
            xB,
            xC,

            p1_pos,
            p1_load,

            p2_load,

            p3_pos,
            p3_load
        )

        # ----------------------------------------------------
        # SUMMARY
        # ----------------------------------------------------

        summary = f"""
# ✅ Calculation Completed

## Mathematical Model

### AX = B

Unknown vector:

### X = [Rₐ, Rᵦ, R𝒸]ᵀ

---

## Matrix Determinant

**det(A) = {np.linalg.det(A):.4f}**

Because det(A) ≠ 0:

**A is invertible → Unique solution exists**

---

## Before Truck Movement

| Support | Reaction |
|---|---:|
| A | **{before[0]:.3f} kN** |
| B | **{before[1]:.3f} kN** |
| C | **{before[2]:.3f} kN** |

---

## After Truck Movement

Truck moved to:

**{truck_position:.2f} m**

| Support | Reaction |
|---|---:|
| A | **{after[0]:.3f} kN** |
| B | **{after[1]:.3f} kN** |
| C | **{after[2]:.3f} kN** |

---

## Verification

Maximum error in AX = B:

**{error:.10e}**

A value close to zero confirms the solution.
"""

        return (

            summary,

            A_text,
            B_text,
            inverse_text,

            results,

            bridge_fig,
            comparison_fig,
            movement_fig
        )

    except Exception as e:

        return (

            f"""
# ❌ ERROR

{str(e)}

Please check that:

• A < B < C
• Load positions are valid
• Loads are positive numbers
""",

            "",
            "",
            "",

            pd.DataFrame(),

            None,
            None,
            None
        )


# ============================================================
# GRADIO FRONTEND
# ============================================================

with gr.Blocks(
    title="Bridge Load Distribution Calculator"
) as app:

    # --------------------------------------------------------
    # HEADER
    # --------------------------------------------------------

    gr.Markdown(
        """
# 🏗️ BRIDGE LOAD DISTRIBUTION CALCULATOR

### B.Tech Mathematics Mini Project — A2

**AX = B | Matrix Inverse | Unique Solution**

Calculate the reaction forces at **three supports A, B and C**
and study how the reactions change when the truck moves.
"""
    )

    # --------------------------------------------------------
    # SUPPORT INPUTS
    # --------------------------------------------------------

    gr.Markdown(
        "## 🌉 Support Positions"
    )

    with gr.Row():

        xA = gr.Number(
            value=0,
            label="Support A Position (m)"
        )

        xB = gr.Number(
            value=5,
            label="Support B Position (m)"
        )

        xC = gr.Number(
            value=10,
            label="Support C Position (m)"
        )

    # --------------------------------------------------------
    # LOAD 1
    # --------------------------------------------------------

    gr.Markdown(
        "## 📦 Load P1"
    )

    with gr.Row():

        p1_pos = gr.Number(
            value=2,
            label="P1 Position (m)"
        )

        p1_load = gr.Number(
            value=20,
            label="P1 Load (kN)"
        )

    # --------------------------------------------------------
    # LOAD 2
    # --------------------------------------------------------

    gr.Markdown(
        "## 📦 Load P2 / Truck"
    )

    with gr.Row():

        p2_pos = gr.Number(
            value=5,
            label="P2 Initial Position (m)"
        )

        p2_load = gr.Number(
            value=30,
            label="P2 Load (kN)"
        )

    # --------------------------------------------------------
    # LOAD 3
    # --------------------------------------------------------

    gr.Markdown(
        "## 📦 Load P3"
    )

    with gr.Row():

        p3_pos = gr.Number(
            value=8,
            label="P3 Position (m)"
        )

        p3_load = gr.Number(
            value=25,
            label="P3 Load (kN)"
        )

    # --------------------------------------------------------
    # TRUCK
    # --------------------------------------------------------

    gr.Markdown(
        "## 🚚 Truck Movement"
    )

    truck_position = gr.Slider(

        minimum=0.5,
        maximum=9.5,

        value=7,

        step=0.5,

        label="New Truck Position (m)"
    )

    calculate_button = gr.Button(

        "🔵 CALCULATE REACTIONS",

        variant="primary"
    )

    # --------------------------------------------------------
    # RESULTS
    # --------------------------------------------------------

    gr.Markdown(
        "---"
    )

    summary = gr.Markdown()

    with gr.Row():

        A_output = gr.Textbox(
            label="Matrix A",
            lines=6
        )

        B_output = gr.Textbox(
            label="Vector B",
            lines=6
        )

        inverse_output = gr.Textbox(
            label="Matrix A⁻¹",
            lines=10
        )

    results_table = gr.Dataframe(
        label="Reaction Force Comparison",
        interactive=False
    )

    # --------------------------------------------------------
    # CHARTS
    # --------------------------------------------------------

    gr.Markdown(
        "## 📊 Visual Results"
    )

    bridge_plot = gr.Plot(
        label="Bridge Diagram"
    )

    comparison_plot = gr.Plot(
        label="Reaction Comparison"
    )

    movement_plot = gr.Plot(
        label="Truck Movement Analysis"
    )

    # --------------------------------------------------------
    # BUTTON ACTION
    # --------------------------------------------------------

    calculate_button.click(

        fn=calculate_bridge,

        inputs=[

            xA,
            xB,
            xC,

            p1_pos,
            p1_load,

            p2_pos,
            p2_load,

            p3_pos,
            p3_load,

            truck_position
        ],

        outputs=[

            summary,

            A_output,
            B_output,
            inverse_output,

            results_table,

            bridge_plot,
            comparison_plot,
            movement_plot
        ]
    )


# ============================================================
# LAUNCH
# ============================================================

app.launch(
    share=True
)
```
